# Converted the Amenities column into binary (0/1) features using multi-label binarizer, where each unique amenity is represented by a separate column indicating its presence or absence."

```python
import pandas as pd
import ast
from sklearn.preprocessing import MultiLabelBinarizer


df['Amenities'] = df['Amenities'].apply(ast.literal_eval)


mlb = MultiLabelBinarizer()

amenities_df = pd.DataFrame(
    mlb.fit_transform(df['Amenities']),
    columns=mlb.classes_,
    index=df.index
)


df = pd.concat([df.drop('Amenities', axis=1), amenities_df], axis=1)

print(df.head())
```

# Outliers Detection  and Removal 

```python


import matplotlib.pyplot as plt

plt.boxplot(df['Price'])
plt.show()
```

```python
df['Price'].describe()
```

```python
Q1 = df['Price'].quantile(0.25)
Q3 = df['Price'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df = df[(df['Price'] >= lower) & (df['Price'] <= upper)]
```

```python
import matplotlib.pyplot as plt

plt.boxplot(df['Price'])
plt.show()
```

```python
corr = df.corr(numeric_only=True)['Price']
print(corr.sort_values(ascending=False))
```
